```python
#Chroma_db
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
# from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=TextLoader("D:\\Langchain\\speech.txt")
data=loader.load()
data
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=30)
splits=text_splitter.split_documents(data)
```

```python
embidding=HuggingFaceEmbeddings()
vectordb=Chroma.from_documents(documents=splits,embedding=embidding)
vectordb
```

```python
#query it
query="How does the speaker describe the desired outcome of the war?"
docs=vectordb.similarity_search(query)
docs[0].page_content
```

```python
## Saving
vectordb=Chroma.from_documents(documents=splits,embedding=embidding,persist_directory="./chroma_db")
```

```python
db2=Chroma(persist_directory="./chroma_db",embedding_function=embidding)
docs=db2.similarity_search(query)
print(docs[0].page_content)
```

```python
docs=db2.similarity_search_with_score(query)
docs
```

```python
retriever=vectordb.as_retriever()
docs=retriever.invoke(query)
docs[0].page_content
```
